## Data Splitting Methods

To evaluate the model effectively while preventing data leakage, we consider three distinct strategies for partitioning the data into Training (75%), Validation (15%), and Test (10%) sets:

1. **Naive ID-Based Split:** Randomly partition all unique voter IDs into three sets. A pair (id1, id2) is assigned to a split only if both IDs belong to that specific split. This guarantees strict isolation but discards approximately 40% of all pair data.

2. **Duplicate-Centric Component Split:** Construct a graph using only Duplicate (DPL) pairs to identify identity clusters (connected components). Partition these clusters into splits. This achieves 100% retention of duplicate pairs, though we still lose roughly 40% of cross-split non-duplicate pairs.

3. **Identity-Aware Negative Sampling:** Follow the same process as Method 2 for Duplicate data to ensure 100% positive retention, but assume that any pair of IDs from different identity clusters can be treated as a valid non-duplicate negative example. This maximizes data utility while maintaining strict identity isolation.

### Naive ID-Based Split

```python
# Train/val/test split of IDs

from sklearn.model_selection import train_test_split

unique_ids = df_ncvoters['id'].unique()
train_ids, temp_ids = train_test_split(unique_ids, test_size=0.25, random_state=42)
val_ids, test_ids = train_test_split(temp_ids, test_size=0.40, random_state=42)
```

A pair is assigned to a specific split (train, validation, or test) if and only if both voter IDs in that pair belong to that split's corresponding ID set. Pairs with IDs spanning different splits are excluded to prevent leakage.

```python
train_ids_set = set(train_ids)
val_ids_set = set(val_ids)
test_ids_set = set(test_ids)

dup_train = DPL[DPL['id1'].isin(train_ids_set) & DPL['id2'].isin(train_ids_set)]
dup_val = DPL[DPL['id1'].isin(val_ids_set) & DPL['id2'].isin(val_ids_set)]
dup_test = DPL[DPL['id1'].isin(test_ids_set) & DPL['id2'].isin(test_ids_set)]

ndup_train = NDPL[NDPL['id1'].isin(train_ids_set) & NDPL['id2'].isin(train_ids_set)]
ndup_val = NDPL[NDPL['id1'].isin(val_ids_set) & NDPL['id2'].isin(val_ids_set)]
ndup_test = NDPL[NDPL['id1'].isin(test_ids_set) & NDPL['id2'].isin(test_ids_set)]
```

```python
total_dpl = len(DPL)
total_ndpl = len(NDPL)

# Calculate sums of splits
dup_kept = len(dup_train) + len(dup_val) + len(dup_test)
ndup_kept = len(ndup_train) + len(ndup_val) + len(ndup_test)

# Calculate discarded (pairs that cross splits)
dup_discarded = total_dpl - dup_kept
ndup_discarded = total_ndpl - ndup_kept

print("=== Split Verification ===")
print(f"Original DPL count: {total_dpl}")
print(f"  - Train: {len(dup_train)}")
print(f"  - Val:   {len(dup_val)}")
print(f"  - Test:  {len(dup_test)}")
print(f"  - Kept:  {dup_kept} ({dup_kept/total_dpl:.2%})")
print(f"  - Discarded (cross-split): {dup_discarded} ({dup_discarded/total_dpl:.2%})")

print(f"\nOriginal NDPL count: {total_ndpl}")
print(f"  - Train: {len(ndup_train)}")
print(f"  - Val:   {len(ndup_val)}")
print(f"  - Test:  {len(ndup_test)}")
print(f"  - Kept:  {ndup_kept} ({ndup_kept/total_ndpl:.2%})")
print(f"  - Discarded (cross-split): {ndup_discarded} ({ndup_discarded/total_ndpl:.2%})")
```

### Duplicate-Centric Component Split:

We will first build a graph where connected components represents duplicates of a single person and divide these components into train/test/val. We will then do a Naive split on the non-duplicates based on nodes of duplicate split

```python
import networkx as nx
import pandas as pd
import random

G = nx.Graph()
G.add_edges_from(zip(DPL['id1'], DPL['id2']))
connected_components = list(nx.connected_components(G))

component_edges = []
for comp in connected_components:
    component_edges.append((comp, G.subgraph(comp).number_of_edges()))

random.seed(42)
random.shuffle(component_edges)

total_edges = G.number_of_edges()
current_edges = 0
train_nodes, val_nodes, test_nodes = set(), set(), set()

for comp, n_edges in component_edges:
    if current_edges < 0.75 * total_edges:
        train_nodes.update(comp)
    elif current_edges < 0.90 * total_edges:
        val_nodes.update(comp)
    else:
        test_nodes.update(comp)
    current_edges += n_edges
```

```python
dup_train_comp = DPL[DPL['id1'].isin(train_nodes) & DPL['id2'].isin(train_nodes)].copy()
dup_val_comp = DPL[DPL['id1'].isin(val_nodes) & DPL['id2'].isin(val_nodes)].copy()
dup_test_comp = DPL[DPL['id1'].isin(test_nodes) & DPL['id2'].isin(test_nodes)].copy()

ndup_train_comp = NDPL[NDPL['id1'].isin(train_nodes) & NDPL['id2'].isin(train_nodes)].copy()
ndup_val_comp = NDPL[NDPL['id1'].isin(val_nodes) & NDPL['id2'].isin(val_nodes)].copy()
ndup_test_comp = NDPL[NDPL['id1'].isin(test_nodes) & NDPL['id2'].isin(test_nodes)].copy()
```

### Identity-Aware Negative Sampling:

Since we loose non duplicates in method 2. We will assume that any two entries in different clusters are non duplicates and we will produce these non duplicates by hand to match the 10:1 ratio of the original data

```python
id_to_cluster = {}
for cluster_id, component in enumerate(connected_components):
    for voter_id in component:
        id_to_cluster[voter_id] = cluster_id

all_voter_ids = df_ncvoters['id'].unique()
next_id = len(connected_components)
for v_id in all_voter_ids:
    if v_id not in id_to_cluster:
        id_to_cluster[v_id] = next_id
        next_id += 1
```

```python
def generate_negatives(id_set, target_count, cluster_map, seed=42):
    import random
    random.seed(seed)
    id_list = list(id_set)
    neg_list = []
    seen = set()
    while len(neg_list) < target_count:
        id1, id2 = random.sample(id_list, 2)
        pair = tuple(sorted((id1, id2)))
        if pair not in seen and cluster_map[id1] != cluster_map[id2]:
            neg_list.append({'id1': id1, 'id2': id2, 'label': 0})
            seen.add(pair)
    return pd.DataFrame(neg_list)
```

```python
dup_train_m3 = DPL[DPL['id1'].isin(train_nodes) & DPL['id2'].isin(train_nodes)].copy()
dup_val_m3 = DPL[DPL['id1'].isin(val_nodes) & DPL['id2'].isin(val_nodes)].copy()
dup_test_m3 = DPL[DPL['id1'].isin(test_nodes) & DPL['id2'].isin(test_nodes)].copy()

for df in [dup_train_m3, dup_val_m3, dup_test_m3]: df['label'] = 1

ndup_train_m3 = generate_negatives(train_nodes, len(dup_train_m3) * 10, id_to_cluster, seed=42)
ndup_val_m3 = generate_negatives(val_nodes, len(dup_val_m3) * 10, id_to_cluster, seed=43)
ndup_test_m3 = generate_negatives(test_nodes, len(dup_test_m3) * 10, id_to_cluster, seed=44)
```
